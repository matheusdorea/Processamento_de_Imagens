```python
img_bgr = cv2.imread('pokemon.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
img_rgb
```

```python
def quantizar_imagem(img, bits):
    n_niveis = 2**bits
    # Normaliza para [0, n_niveis - 1], arredonda e reescala para [0, 255]
    img_quantizada = np.floor((img / 256.0) * n_niveis) * (255.0 / (n_niveis - 1))
    return img_quantizada.astype(np.uint8)
```

```python
img_1bit = quantizar_imagem(img_gray, 1)  # 2 níveis
img_2bits = quantizar_imagem(img_gray, 2)  # 4 níveis
img_4bits = quantizar_imagem(img_gray, 4)  # 16 níveis
img_8bits = quantizar_imagem(img_gray, 8)  # 256 níveis
```

```python
fig, ax = plt.subplots(2, 2, figsize=(8, 6))
#RGB
ax[0, 0].imshow(img_1bit, cmap='gray')
ax[0, 0].set_title("1 Bit")
ax[0, 0].axis('off')

#BGR
ax[0, 1].imshow(img_2bits, cmap='gray')
ax[0, 1].set_title("2 Bits")
ax[0, 1].axis('off')

#HSV
ax[1, 0].imshow(img_4bits, cmap='gray')
ax[1, 0].set_title("4 Bits")
ax[1, 0].axis('off')

#GRAY
ax[1, 1].imshow(img_8bits, cmap='gray')
ax[1, 1].set_title("8 Bits")
ax[1, 1].axis('off')

plt.tight_layout()
plt.savefig('quantizacao')
plt.show()
```

R = Ao diminuir a quantidade de bits de quantização, menos detalhes aparecem, pois temos menos "níveis de cinza" após a normalização dos pixels.
